```python
"""
Question 2: Customer Purchase Analysis

Using the provided dataset 'customer_purchases.csv', perform the following analyses:

1. Calculate the following metrics per customer:
   - Total amount spent
   - Average purchase value
   - Number of purchases
   - Most frequently bought category

2. Create a summary DataFrame with:
   - Top 5 customers by total spend
   - Bottom 5 customers by total spend

3. Calculate the monthly purchase trends:
   - Total sales per month
   - Average purchase value per month

Bonus: Identify any customers who haven't made a purchase in the last 3 months
"""

# Your code here
```

```python
import pandas as pd
import scipy.stats as stats
import numpy as np
```

### 1. Metrics per customer

```python
df = pd.read_csv('../data/customer_purchases.csv')
df.info()
df.describe()
df
```

```python
t_df = df.copy()
```

```python
# To make sure we can get monthly data later, first we convert purchase date to datetime
t_df['purchase_date'] = pd.to_datetime(t_df['purchase_date'])
t_df.info()
```

```python
# Total spent per customer
per_customer_data = t_df.groupby('customer_id')
total_spent = per_customer_data['amount'].sum()
total_spent
```

```python
# Averae purchase value per customer
avg_purchase_per_customer = per_customer_data['amount'].mean()
avg_purchase_per_customer
```

```python
# Number of purchases per customer
n_purchases_per_customer = per_customer_data.describe()['purchase_id']['count']
n_purchases_per_customer
```

```python
# Most frequent catagory of purchase per customer
most_freq_cat = per_customer_data.agg(mfc=('category', lambda x: x.mode()[0]))
# Tried doing it a simpler way, like the other queries, but couldn't make it clean enough
# Read about .agg on the docs
most_freq_cat
```

### 2. Summary DataFrame

```python
summary_df = pd.DataFrame()
```

```python
# Top 5 customers by total spend
top5 = total_spent.nlargest(5)

# Bottom 5 customer by total spend
bottom5 = total_spent.nsmallest(5)
```

```python
summary_df = pd.concat([top5, bottom5])
summary_df
```

### 3. Monthly purchase trends

```python
# Total sales per month (number of sales. Not amount sold)
# We dont have a standalone "month" columns, so it would be helpful and appropriate to create one
t_df['month'] = t_df['purchase_date'].dt.to_period('M')

sales_month = t_df.groupby('month')
sales_month.describe()['purchase_id']['count']
```

```python
# Expanding a little bit:
# Total made in sales per month
sales_month['amount'].sum()
```

```python
# Average puschase value per month

sales_month.describe()
sales_month['amount'].mean()
```
